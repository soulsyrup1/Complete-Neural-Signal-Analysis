<h2 style="color:#E8E8E8;margin:0;">Per-Channel MFDFA Fluctuation Curves for All q Values</h2>
<p style="margin:8px 0 0 0;">This section visualizes the fluctuation functions \(F_q(s)\) from a completed MFDFA run. The previous MFDFA code already computed the numerical quantities. This code does not recompute the full MFDFA from the raw EEG. Instead, it loads the saved MFDFA result file and makes one detailed log-log fluctuation plot for each EEG channel, showing all available \(q\)-dependent fluctuation curves.</p>
</div>
<div style="background-color:#222222;border-left:4px solid #D0D0D0;padding:12px 14px;border-radius:6px;margin-bottom:18px;">
<h3 style="color:#E8E8E8;margin-top:0;">The Central Idea</h3>
<p style="margin-bottom:0;">MFDFA asks how fluctuations grow as the observation scale grows. If a signal is examined using short windows, medium windows, and long windows, the fluctuation size \(F_q(s)\) usually changes with the scale \(s\). The shape of this relationship tells us whether the signal has simple scaling or richer multifractal scaling. In these plots, every curve is one \(q\) value. Together, the curves show whether small, typical, and large fluctuations scale in the same way or in different ways.</p>
</div>
<div style="display:flex;flex-wrap:wrap;gap:18px;">
<div style="flex:1;min-width:320px;">
<h3 style="color:#E8E8E8;margin-top:0;">1. What This Code Loads</h3>
<p>The code loads a saved NumPy archive called <code>mfdfa_results.npz</code>. This archive contains the MFDFA results produced earlier. It includes the channel results, the tested \(q\) values, and the tested scale values.</p>
<p>Each channel result contains the fluctuation matrix \(F_q(s)\). In this matrix, one axis corresponds to scale \(s\), and the other axis corresponds to \(q\). The code uses this matrix to draw all fluctuation curves for each channel.</p>
<p>Because this code uses saved results, it is a visualization and interpretation stage. The heavy MFDFA calculation has already been done.</p>
<h3 style="color:#E8E8E8;">2. What \(F_q(s)\) Means</h3>
<p>The fluctuation function \(F_q(s)\) tells us how large the detrended fluctuations are at scale \(s\), after weighting those fluctuations according to \(q\).</p>
<p>A small scale \(s\) looks at short-time structure. A large scale \(s\) looks at slower, broader structure. If \(F_q(s)\) increases smoothly as \(s\) increases, the signal has scale-dependent organization.</p>
<p>The important relationship is:</p>
<p style="text-align:center;">\(F_q(s)\sim s^{h(q)}\)</p>
<p>On a log-log plot, this relationship becomes approximately linear:</p>
<p style="text-align:center;">\(\log(F_q(s))\approx h(q)\log(s)+C\)</p>
<p>The slope is the generalized Hurst exponent \(h(q)\). Therefore, these plots let us visually inspect the scaling behavior that supports the numerical \(h(q)\) estimates.</p>
</div>
<div style="flex:1;min-width:320px;">
<h3 style="color:#E8E8E8;margin-top:0;">3. What the q Values Do</h3>
<p>The \(q\) parameter changes which parts of the fluctuation distribution receive more attention. Negative \(q\) values emphasize smaller fluctuations. Positive \(q\) values emphasize larger fluctuations. Values near zero emphasize more typical fluctuations through a logarithmic average.</p>
<p>If all \(q\)-curves are nearly parallel, then small and large fluctuations scale in similar ways. That is closer to simple scaling. If the \(q\)-curves separate strongly, bend differently, or show different slopes, then small and large fluctuations behave differently across scale. That is the visual signature of multifractal behavior.</p>
<h3 style="color:#E8E8E8;">4. Why Log-Log Plots Are Used</h3>
<p>Power-law scaling is hard to see on ordinary axes. A curve that follows \(F_q(s)\sim s^{h(q)}\) becomes a straight line when both axes are logarithmic. That is why these plots use log-log axes.</p>
<p>A good scaling region appears as a clean, nearly straight line. A poor scaling region appears curved, broken, noisy, or dominated by one scale. The numerical slopes may still exist, but the visual inspection tells us whether the slopes should be trusted.</p>
</div>
</div>
<div style="margin-top:18px;border-top:1px solid #404040;padding-top:16px;">
<h3 style="color:#E8E8E8;margin-top:0;">5. Why One Plot per Channel Is Useful</h3>
<p>The full MFDFA summary gives compact metrics such as alpha width, \(h(q)\), fit quality, and spectrum asymmetry. Those summaries are useful, but they compress a large amount of information. A per-channel \(F_q(s)\) plot shows the raw scaling curves behind those summaries.</p>
<p>This is like inspecting the foundation underneath a numerical result. A channel can have a large alpha width, but if the fluctuation curves are jagged or curved, the result should be treated carefully. Conversely, a channel with clean, well-separated, nearly straight curves gives stronger evidence for structured multifractal scaling.</p>
</div>
<div style="display:flex;flex-wrap:wrap;gap:18px;margin-top:12px;">
<div style="flex:1;min-width:320px;">
<h3 style="color:#E8E8E8;">6. What the Individual Channel Plots Show</h3>
<p>Each individual plot contains all \(q\)-dependent fluctuation curves for one EEG channel. The x-axis is scale \(s\), and the y-axis is \(F_q(s)\). The colorbar shows which color corresponds to which \(q\) value.</p>
<p>These plots are used to check whether the fluctuation curves are smooth, monotonic, separated, parallel, curved, or unstable. They are especially valuable when a channel appears important in the MFDFA summary table.</p>
<h3 style="color:#E8E8E8;">7. What the Contact Sheet Shows</h3>
<p>The contact sheet places all channels into one large grid of small plots. It is not meant for fine reading of every curve. Its purpose is comparison. It lets us see quickly whether all channels have similar scaling behavior or whether some channels are visibly different.</p>
<p>If one channel looks unusual in the contact sheet, it can then be inspected in its full-size individual plot.</p>
</div>
<div style="flex:1;min-width:320px;">
<h3 style="color:#E8E8E8;">8. How to Interpret Curve Separation</h3>
<p>If the curves for different \(q\) values lie close together and have similar slopes, the signal has relatively uniform scaling behavior. If the curves separate strongly or change slope depending on \(q\), the signal behaves differently for small versus large fluctuations.</p>
<p>This curve separation is one of the most intuitive visual signs of multifractality. However, separation alone is not enough. The curves should also be smooth and approximately linear on log-log axes.</p>
<h3 style="color:#E8E8E8;">9. How to Interpret Curvature</h3>
<p>Curvature in the log-log fluctuation curves means that one straight scaling slope may not describe all scales equally well. This can occur because of true scale transitions, noise, artifacts, too few scales, or inappropriate scale limits.</p>
<p>Therefore, a channel with strong curvature should be interpreted cautiously. The numerical MFDFA metrics may still be informative, but they may summarize a more complicated scale structure than a single slope can fully capture.</p>
</div>
</div>
<div style="margin-top:18px;background-color:#222222;border-left:4px solid #D0D0D0;padding:12px 14px;border-radius:6px;">
<h3 style="color:#E8E8E8;margin-top:0;">Core Intuition</h3>
<p style="margin-bottom:0;">These plots are the visual inspection stage of MFDFA. They show how \(F_q(s)\) changes with scale for every \(q\) value. Clean, nearly straight, well-separated curves support reliable multifractal interpretation. Curved, noisy, broken, or scale-dominated curves warn us to be cautious. The safest conclusion comes when the visual fluctuation curves agree with the numerical MFDFA summaries.</p>
</div>
</div>

```python
import os
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm, colors

try:
    from IPython.display import display, Markdown
    CAN_DISPLAY_MARKDOWN = True
except Exception:
    display = None
    Markdown = None
    CAN_DISPLAY_MARKDOWN = False

base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
mfdfa_dir = os.path.join(base_dir, "results", "mfdfa_manual")
npz_path = os.path.join(mfdfa_dir, "mfdfa_results.npz")

out_dir = os.path.join(mfdfa_dir, "plots", "per_channel_all_41_Fq")
os.makedirs(out_dir, exist_ok=True)

summary_csv_path = os.path.join(out_dir, "per_channel_all_q_Fq_visual_summary.csv")
report_txt_path = os.path.join(out_dir, "per_channel_all_q_Fq_interpretation_report.md")

ACCENT = "cyan"
BG = "black"
WHITE = "#E8FFFF"
eps = 1e-12

SHOW_INDIVIDUAL_PLOTS_IN_NOTEBOOK = True
SHOW_CONTACT_SHEET_IN_NOTEBOOK = True

SAVE_INDIVIDUAL_PLOTS = True
SAVE_CONTACT_SHEET = True

CONTACT_SHEET_NCOLS = 4
CONTACT_SHEET_DPI = 180
INDIVIDUAL_PLOT_DPI = 180

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
    "legend.facecolor": BG,
    "legend.edgecolor": ACCENT,
})

def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)

def print_plot_description(title, what_you_see, patterns_to_look_for, cautions=None):
    print("\n" + "=" * 96)
    print(title)
    print("=" * 96)
    print("\nWhat you are seeing:")
    for item in what_you_see:
        print(f"  - {item}")
    print("\nPatterns to look for:")
    for item in patterns_to_look_for:
        print(f"  - {item}")
    if cautions:
        print("\nCautions:")
        for item in cautions:
            print(f"  - {item}")
    print("=" * 96 + "\n")

def fmt(x, digits=4):
    try:
        if x is None or not np.isfinite(x):
            return "not available"
        return f"{x:.{digits}f}"
    except Exception:
        return str(x)

def bullet_rows(df, name_col, value_cols, n=10):
    if df is None or len(df) == 0:
        return "- No values available."

    lines = []

    for _, row in df.head(n).iterrows():
        name = row.get(name_col, "unknown")
        pieces = []

        for c in value_cols:
            if c in row:
                pieces.append(f"{c}={fmt(row[c])}")

        lines.append(f"- **{name}**: " + ", ".join(pieces))

    return "\n".join(lines)

def unwrap_result_item(item):
    if isinstance(item, np.ndarray) and item.shape == ():
        return item.item()
    return item

def safe_log_fit(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & (x > 0) & np.isfinite(y) & (y > 0)

    if np.sum(mask) < 3:
        return np.nan, np.nan, np.nan, int(np.sum(mask))

    lx = np.log(x[mask])
    ly = np.log(y[mask])

    coef = np.polyfit(lx, ly, 1)
    yhat = coef[0] * lx + coef[1]

    ss_res = float(np.sum((ly - yhat) ** 2))
    ss_tot = float(np.sum((ly - np.mean(ly)) ** 2))
    r2 = 1.0 - ss_res / (ss_tot + eps)

    return float(coef[0]), float(coef[1]), float(r2), int(np.sum(mask))

def summarize_channel_curves(channel, scales, Fq, q_vals):
    scales = np.asarray(scales, dtype=float)
    Fq = np.asarray(Fq, dtype=float)
    q_vals = np.asarray(q_vals, dtype=float)

    slopes = []
    r2s = []
    valid_counts = []
    log_ranges = []
    first_values = []
    last_values = []

    for qi, q in enumerate(q_vals):
        y = Fq[:, qi]
        mask = np.isfinite(y) & (y > 0) & np.isfinite(scales) & (scales > 0)

        if np.sum(mask) >= 3:
            slope, intercept, r2, n_fit = safe_log_fit(scales[mask], y[mask])
            yy = y[mask]

            slopes.append(slope)
            r2s.append(r2)
            valid_counts.append(n_fit)
            log_ranges.append(float(np.nanmax(np.log(yy + eps)) - np.nanmin(np.log(yy + eps))))
            first_values.append(float(yy[0]))
            last_values.append(float(yy[-1]))

        else:
            slopes.append(np.nan)
            r2s.append(np.nan)
            valid_counts.append(int(np.sum(mask)))
            log_ranges.append(np.nan)
            first_values.append(np.nan)
            last_values.append(np.nan)

    slopes = np.asarray(slopes, dtype=float)
    r2s = np.asarray(r2s, dtype=float)
    valid_counts = np.asarray(valid_counts, dtype=float)
    log_ranges = np.asarray(log_ranges, dtype=float)
    first_values = np.asarray(first_values, dtype=float)
    last_values = np.asarray(last_values, dtype=float)

    finite_slopes = slopes[np.isfinite(slopes)]
    finite_r2s = r2s[np.isfinite(r2s)]

    q_min_idx = int(np.nanargmin(q_vals))
    q_max_idx = int(np.nanargmax(q_vals))
    q_zero_idx = int(np.nanargmin(np.abs(q_vals)))

    if np.isfinite(slopes[q_min_idx]) and np.isfinite(slopes[q_max_idx]):
        slope_q_span = float(slopes[q_max_idx] - slopes[q_min_idx])
        slope_q_abs_span = abs(slope_q_span)
    else:
        slope_q_span = np.nan
        slope_q_abs_span = np.nan

    if np.isfinite(first_values[q_min_idx]) and np.isfinite(first_values[q_max_idx]):
        first_curve_separation = float(abs(np.log(first_values[q_max_idx] + eps) - np.log(first_values[q_min_idx] + eps)))
    else:
        first_curve_separation = np.nan

    if np.isfinite(last_values[q_min_idx]) and np.isfinite(last_values[q_max_idx]):
        last_curve_separation = float(abs(np.log(last_values[q_max_idx] + eps) - np.log(last_values[q_min_idx] + eps)))
    else:
        last_curve_separation = np.nan

    if np.isfinite(first_curve_separation) and np.isfinite(last_curve_separation):
        separation_change = float(last_curve_separation - first_curve_separation)
    else:
        separation_change = np.nan

    return {
        "channel": channel,
        "n_q_total": int(len(q_vals)),
        "n_q_with_valid_fit": int(np.sum(np.isfinite(slopes))),
        "mean_curve_slope": float(np.nanmean(slopes)) if finite_slopes.size else np.nan,
        "median_curve_slope": float(np.nanmedian(slopes)) if finite_slopes.size else np.nan,
        "min_curve_slope": float(np.nanmin(slopes)) if finite_slopes.size else np.nan,
        "max_curve_slope": float(np.nanmax(slopes)) if finite_slopes.size else np.nan,
        "curve_slope_range": float(np.nanmax(slopes) - np.nanmin(slopes)) if finite_slopes.size >= 2 else np.nan,
        "slope_qmax_minus_qmin": slope_q_span,
        "abs_slope_qmax_minus_qmin": slope_q_abs_span,
        "slope_qmin": float(slopes[q_min_idx]) if np.isfinite(slopes[q_min_idx]) else np.nan,
        "slope_q0": float(slopes[q_zero_idx]) if np.isfinite(slopes[q_zero_idx]) else np.nan,
        "slope_qmax": float(slopes[q_max_idx]) if np.isfinite(slopes[q_max_idx]) else np.nan,
        "mean_fit_r2": float(np.nanmean(r2s)) if finite_r2s.size else np.nan,
        "median_fit_r2": float(np.nanmedian(r2s)) if finite_r2s.size else np.nan,
        "min_fit_r2": float(np.nanmin(r2s)) if finite_r2s.size else np.nan,
        "mean_valid_scale_points": float(np.nanmean(valid_counts)) if len(valid_counts) else np.nan,
        "mean_log_curve_range": float(np.nanmean(log_ranges)) if np.any(np.isfinite(log_ranges)) else np.nan,
        "first_curve_separation_qmax_qmin_log": first_curve_separation,
        "last_curve_separation_qmax_qmin_log": last_curve_separation,
        "separation_change_from_small_to_large_scale": separation_change,
    }

def describe_slope_range(x):
    if not np.isfinite(x):
        return "The slope range is unavailable."
    if x < 0.03:
        return "The q-dependent slopes are very similar, suggesting weak visible multifractal separation in the fluctuation curves."
    if x < 0.10:
        return "The q-dependent slopes differ modestly, suggesting some multifractal separation."
    if x < 0.25:
        return "The q-dependent slopes differ noticeably, suggesting meaningful multifractal separation."
    return "The q-dependent slopes differ strongly, suggesting broad multifractal separation or possible scale-fit instability that should be inspected visually."

def describe_fit_quality(x):
    if not np.isfinite(x):
        return "Fit quality is unavailable."
    if x >= 0.98:
        return "The log-log fluctuation curves are very close to linear on average."
    if x >= 0.95:
        return "The log-log fluctuation curves are mostly linear on average."
    if x >= 0.90:
        return "The log-log fluctuation curves are usable but should be interpreted carefully."
    return "The log-log fluctuation curves show weak linearity on average, so interpretations should be cautious."

def describe_separation_change(x):
    if not np.isfinite(x):
        return "The separation change is unavailable."
    if abs(x) < 0.05:
        return "The separation between low-q and high-q curves stays almost constant across scale."
    if x > 0:
        return "The separation between low-q and high-q curves increases at larger scales."
    return "The separation between low-q and high-q curves decreases at larger scales."

if not os.path.exists(npz_path):
    raise FileNotFoundError(f"Missing saved MFDFA results: {npz_path}")

data = np.load(npz_path, allow_pickle=True)

results = data["results"]
q_vals = np.asarray(data["q_vals"], dtype=float)
scales_global = np.asarray(data["scales"], dtype=float)

print(f"Loaded {len(results)} channel results")
print(f"q range: {q_vals.min()} to {q_vals.max()} ({len(q_vals)} values)")
print(f"Saved plots to: {out_dir}")

cmap = cm.get_cmap("viridis", len(q_vals))
norm = colors.Normalize(vmin=q_vals.min(), vmax=q_vals.max())

summary_rows = []

for item in results:
    item = unwrap_result_item(item)

    channel = item["channel"]
    scales = np.asarray(item["scales"], dtype=float)
    Fq = np.asarray(item["Fq"], dtype=float)

    row = summarize_channel_curves(channel, scales, Fq, q_vals)

    if "hq" in item:
        hq = np.asarray(item["hq"], dtype=float)

        if np.any(np.isfinite(hq)):
            row["saved_hq_mean"] = float(np.nanmean(hq))
            row["saved_hq_range"] = float(np.nanmax(hq) - np.nanmin(hq)) if np.sum(np.isfinite(hq)) >= 2 else np.nan
        else:
            row["saved_hq_mean"] = np.nan
            row["saved_hq_range"] = np.nan

    if "alpha_q" in item and "f_alpha_q" in item:
        alpha = np.asarray(item["alpha_q"], dtype=float)
        f_alpha = np.asarray(item["f_alpha_q"], dtype=float)
        mask = np.isfinite(alpha) & np.isfinite(f_alpha)

        if np.sum(mask) >= 3:
            row["saved_alpha_width"] = float(np.nanmax(alpha[mask]) - np.nanmin(alpha[mask]))
            row["saved_alpha_min"] = float(np.nanmin(alpha[mask]))
            row["saved_alpha_max"] = float(np.nanmax(alpha[mask]))
        else:
            row["saved_alpha_width"] = np.nan
            row["saved_alpha_min"] = np.nan
            row["saved_alpha_max"] = np.nan

    summary_rows.append(row)

summary_df = pd.DataFrame(summary_rows)
summary_df.to_csv(summary_csv_path, index=False)

print(f"\nSaved visual summary CSV: {summary_csv_path}")

print_plot_description(
    title="PLOT 1 — One Full MFDFA Fluctuation-Curve Plot per Channel",
    what_you_see=[
        "Each figure corresponds to one EEG channel.",
        "Each curve in a figure is one q value.",
        "The x-axis is scale s on a logarithmic axis.",
        "The y-axis is Fq(s) on a logarithmic axis.",
        "The colorbar maps curve color to q value.",
        "Negative q curves emphasize smaller fluctuations.",
        "Positive q curves emphasize larger fluctuations.",
        "Straight lines on log-log axes support power-law scaling.",
    ],
    patterns_to_look_for=[
        "Look for curves that are approximately straight across scales.",
        "Look for curves that bend upward or downward.",
        "Look for strong separation between negative-q and positive-q curves.",
        "Look for curves that are nearly parallel.",
        "Look for curves whose slopes differ strongly across q.",
        "Look for crossings between q curves.",
        "Look for noisy or jagged curves.",
        "Look for channels where only a few scales dominate the visual pattern.",
        "Look for channels where large-q curves behave differently from small-q curves.",
        "Look for channels where low-q curves become unstable.",
        "Look for flattening at small scales.",
        "Look for flattening at large scales.",
        "Look for a clean fan-like structure across q.",
        "Look for one q range that separates from the rest.",
        "Look for channels whose curve pattern is visually different from other channels.",
    ],
    cautions=[
        "A broad fan of curves supports multifractal interpretation only if the curves are also reasonably linear.",
        "Curvature on log-log axes means a single scaling slope may oversimplify the data.",
        "The most important channels from the summary table should be inspected individually.",
    ],
)

individual_paths = []

for item in results:
    item = unwrap_result_item(item)

    channel = item["channel"]
    scales = np.asarray(item["scales"], dtype=float)
    Fq = np.asarray(item["Fq"], dtype=float)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BG)
    style_ax(ax)

    for qi, q in enumerate(q_vals):
        y = Fq[:, qi]
        mask = np.isfinite(y) & (y > 0) & np.isfinite(scales) & (scales > 0)

        if np.sum(mask) >= 2:
            ax.loglog(
                scales[mask],
                y[mask],
                color=cmap(qi),
                linewidth=1.1,
                alpha=0.95
            )

    row = summary_df[summary_df["channel"] == channel]

    if len(row) > 0:
        r = row.iloc[0]
        subtitle = (
            f"slope range={fmt(r['curve_slope_range'], 3)}, "
            f"mean R²={fmt(r['mean_fit_r2'], 3)}"
        )
    else:
        subtitle = ""

    ax.set_title(f"{channel}: All MFDFA Fluctuation Curves\n{subtitle}")
    ax.set_xlabel("Scale s")
    ax.set_ylabel("Fq(s)")

    sm = cm.ScalarMappable(norm=norm, cmap=cmap)
    sm.set_array([])

    cbar = plt.colorbar(sm, ax=ax)
    cbar.set_label("q", color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT)

    for label in cbar.ax.get_yticklabels():
        label.set_color(ACCENT)

    out_path = os.path.join(out_dir, f"{channel}_all_q_Fq_curves.png")

    plt.tight_layout()

    if SAVE_INDIVIDUAL_PLOTS:
        plt.savefig(out_path, dpi=INDIVIDUAL_PLOT_DPI, bbox_inches="tight")

    if SHOW_INDIVIDUAL_PLOTS_IN_NOTEBOOK:
        plt.show()

    plt.close(fig)

    individual_paths.append(out_path)
    print(f"Saved: {out_path}")

print_plot_description(
    title="PLOT 2 — Contact Sheet of All Channels and All q Curves",
    what_you_see=[
        "This plot shows all channel fluctuation-curve plots in one grid.",
        "Each small panel is one channel.",
        "Each curve inside a panel is one q value.",
        "The x-axis is scale s.",
        "The y-axis is Fq(s).",
        "The shared colorbar maps color to q value.",
        "This plot is designed for comparison across channels, not detailed curve reading.",
    ],
    patterns_to_look_for=[
        "Look for channels whose entire panel looks different from the others.",
        "Look for panels with cleaner straight-line scaling.",
        "Look for panels with strong curvature.",
        "Look for panels with strong q-curve separation.",
        "Look for panels where q curves are nearly parallel.",
        "Look for panels where curves cross.",
        "Look for panels where only high-q curves are unstable.",
        "Look for panels where only low-q curves are unstable.",
        "Look for channels with unusually steep curves.",
        "Look for channels with unusually flat curves.",
        "Look for regional similarity if neighboring channel names behave similarly.",
        "Look for isolated visual outliers.",
        "Look for whether all channels share the same broad scaling shape.",
        "Look for whether curve separation grows or shrinks with scale.",
        "Look for panels that should be inspected in full size.",
    ],
    cautions=[
        "Small panels compress detail, so use the individual saved plots for final inspection.",
        "A contact-sheet outlier may reflect signal structure or artifact.",
        "The contact sheet is a screening view, not a final numerical conclusion.",
    ],
)

contact_path = None

if SAVE_CONTACT_SHEET or SHOW_CONTACT_SHEET_IN_NOTEBOOK:
    n_channels = len(results)
    ncols = CONTACT_SHEET_NCOLS
    nrows = int(np.ceil(n_channels / ncols))

    fig, axes = plt.subplots(nrows, ncols, figsize=(16, 4 * nrows), facecolor=BG)
    axes = np.atleast_1d(axes).ravel()

    for ax in axes:
        style_ax(ax)

    for idx, item in enumerate(results):
        item = unwrap_result_item(item)

        channel = item["channel"]
        scales = np.asarray(item["scales"], dtype=float)
        Fq = np.asarray(item["Fq"], dtype=float)

        ax = axes[idx]

        for qi, q in enumerate(q_vals):
            y = Fq[:, qi]
            mask = np.isfinite(y) & (y > 0) & np.isfinite(scales) & (scales > 0)

            if np.sum(mask) >= 2:
                ax.loglog(
                    scales[mask],
                    y[mask],
                    color=cmap(qi),
                    linewidth=0.7,
                    alpha=0.8
                )

        row = summary_df[summary_df["channel"] == channel]

        if len(row) > 0:
            r = row.iloc[0]
            title = f"{channel} | range={fmt(r['curve_slope_range'], 2)}"
        else:
            title = channel

        ax.set_title(title, fontsize=10)
        ax.set_xlabel("s")
        ax.set_ylabel("Fq")

    for j in range(n_channels, len(axes)):
        axes[j].axis("off")

    fig.suptitle("Per-Channel MFDFA Fluctuation Curves, All q Values", color=ACCENT, fontsize=16)

    sm = cm.ScalarMappable(norm=norm, cmap=cmap)
    sm.set_array([])

    cbar = fig.colorbar(sm, ax=axes.tolist(), fraction=0.015, pad=0.01)
    cbar.set_label("q", color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT)

    for label in cbar.ax.get_yticklabels():
        label.set_color(ACCENT)

    contact_path = os.path.join(out_dir, "all_channels_all_q_Fq_contact_sheet.png")

    plt.tight_layout(rect=[0, 0, 1, 0.98])

    if SAVE_CONTACT_SHEET:
        plt.savefig(contact_path, dpi=CONTACT_SHEET_DPI, bbox_inches="tight")

    if SHOW_CONTACT_SHEET_IN_NOTEBOOK:
        plt.show()

    plt.close(fig)

    print(f"\nSaved contact sheet: {contact_path}")

valid_df = summary_df.copy()
valid_df = valid_df[np.isfinite(valid_df["curve_slope_range"])].copy()

top_slope_range_df = valid_df.sort_values("curve_slope_range", ascending=False)
low_slope_range_df = valid_df.sort_values("curve_slope_range", ascending=True)
top_q_span_df = valid_df.sort_values("abs_slope_qmax_minus_qmin", ascending=False)
top_fit_df = valid_df.sort_values("mean_fit_r2", ascending=False)
weak_fit_df = valid_df.sort_values("mean_fit_r2", ascending=True)

top_sep_change_df = valid_df.copy()
top_sep_change_df["abs_separation_change"] = top_sep_change_df["separation_change_from_small_to_large_scale"].abs()
top_sep_change_df = top_sep_change_df.sort_values("abs_separation_change", ascending=False)

if "saved_alpha_width" in valid_df.columns:
    alpha_width_df = valid_df[np.isfinite(valid_df["saved_alpha_width"])].sort_values("saved_alpha_width", ascending=False)
else:
    alpha_width_df = pd.DataFrame()

slope_range_values = valid_df["curve_slope_range"].to_numpy(dtype=float)
r2_values = valid_df["mean_fit_r2"].to_numpy(dtype=float)
slope_q_span_values = valid_df["abs_slope_qmax_minus_qmin"].to_numpy(dtype=float)
sep_change_values = valid_df["separation_change_from_small_to_large_scale"].to_numpy(dtype=float)

mean_slope_range = float(np.nanmean(slope_range_values)) if len(slope_range_values) else np.nan
median_slope_range = float(np.nanmedian(slope_range_values)) if len(slope_range_values) else np.nan
max_slope_range = float(np.nanmax(slope_range_values)) if len(slope_range_values) else np.nan
min_slope_range = float(np.nanmin(slope_range_values)) if len(slope_range_values) else np.nan

mean_r2 = float(np.nanmean(r2_values)) if len(r2_values) else np.nan
median_r2 = float(np.nanmedian(r2_values)) if len(r2_values) else np.nan
min_r2 = float(np.nanmin(r2_values)) if len(r2_values) else np.nan

mean_q_span = float(np.nanmean(slope_q_span_values)) if len(slope_q_span_values) else np.nan
median_q_span = float(np.nanmedian(slope_q_span_values)) if len(slope_q_span_values) else np.nan

mean_sep_change = float(np.nanmean(sep_change_values)) if len(sep_change_values) else np.nan
median_sep_change = float(np.nanmedian(sep_change_values)) if len(sep_change_values) else np.nan

report = ""

report += "# Interpretation of All-q MFDFA Fluctuation-Curve Results\n\n"

report += (
    "This report interprets the saved all-q fluctuation-curve visualizations. "
    "The purpose is to check whether the channel-level MFDFA summaries are supported by the raw log-log fluctuation curves \(F_q(s)\).\n\n"
)

report += "---\n\n"

report += "## 1. What was inspected\n\n"

report += f"- Channels loaded: **{len(results)}**\n"
report += f"- q values loaded: **{len(q_vals)}**\n"
report += f"- q range: **{fmt(np.nanmin(q_vals))} to {fmt(np.nanmax(q_vals))}**\n"
report += f"- Global scale count: **{len(scales_global)}**\n"
report += f"- Individual plot directory: **{out_dir}**\n\n"

report += (
    "Each channel was inspected through all available \(F_q(s)\) curves. "
    "For each q curve, a log-log line was fitted to summarize its slope and fit quality. "
    "These derived quantities are not a replacement for the original MFDFA output, but they are useful for judging whether the visual scaling behavior looks stable.\n\n"
)

report += "---\n\n"

report += "## 2. Overall curve-separation behavior\n\n"

report += f"- Mean q-dependent slope range: **{fmt(mean_slope_range)}**\n"
report += f"- Median q-dependent slope range: **{fmt(median_slope_range)}**\n"
report += f"- Minimum q-dependent slope range: **{fmt(min_slope_range)}**\n"
report += f"- Maximum q-dependent slope range: **{fmt(max_slope_range)}**\n"
report += f"- Mean absolute qmax-qmin slope difference: **{fmt(mean_q_span)}**\n"
report += f"- Median absolute qmax-qmin slope difference: **{fmt(median_q_span)}**\n\n"

report += f"**Interpretation of median slope range:** {describe_slope_range(median_slope_range)}\n\n"

report += (
    "The q-dependent slope range is a visual-support metric. If the slopes of the q curves differ strongly, then small-fluctuation and large-fluctuation scaling are behaving differently. "
    "That supports a multifractal reading, provided the curves are also reasonably linear in log-log space.\n\n"
)

report += "---\n\n"

report += "## 3. Channels with strongest q-dependent separation\n\n"

report += "Largest q-dependent slope ranges:\n\n"
report += bullet_rows(
    top_slope_range_df,
    "channel",
    ["curve_slope_range", "slope_qmin", "slope_q0", "slope_qmax", "mean_fit_r2", "mean_valid_scale_points"],
    n=12,
)
report += "\n\n"

report += "Smallest q-dependent slope ranges:\n\n"
report += bullet_rows(
    low_slope_range_df,
    "channel",
    ["curve_slope_range", "slope_qmin", "slope_q0", "slope_qmax", "mean_fit_r2", "mean_valid_scale_points"],
    n=12,
)
report += "\n\n"

if len(top_slope_range_df) > 0:
    row = top_slope_range_df.iloc[0]
    report += (
        f"The channel with the largest q-dependent slope range is **{row['channel']}** "
        f"(slope range={fmt(row['curve_slope_range'])}, mean R²={fmt(row['mean_fit_r2'])}). "
        f"{describe_slope_range(row['curve_slope_range'])} "
        f"{describe_fit_quality(row['mean_fit_r2'])}\n\n"
    )

if len(low_slope_range_df) > 0:
    row = low_slope_range_df.iloc[0]
    report += (
        f"The channel with the smallest q-dependent slope range is **{row['channel']}** "
        f"(slope range={fmt(row['curve_slope_range'])}). "
        "This channel has the most similar q-curve slopes among the valid channels.\n\n"
    )

report += (
    "A large q-dependent slope range is most meaningful when the corresponding individual channel plot shows clean, nearly straight curves rather than noisy or curved curves.\n\n"
)

report += "---\n\n"

report += "## 4. Log-log fit quality of the Fq curves\n\n"

report += f"- Mean curve-fit R²: **{fmt(mean_r2)}**\n"
report += f"- Median curve-fit R²: **{fmt(median_r2)}**\n"
report += f"- Minimum curve-fit R²: **{fmt(min_r2)}**\n\n"

report += f"**Interpretation of median curve-fit R²:** {describe_fit_quality(median_r2)}\n\n"

report += "Strongest average log-log curve fits:\n\n"
report += bullet_rows(
    top_fit_df,
    "channel",
    ["mean_fit_r2", "min_fit_r2", "curve_slope_range", "slope_q0", "mean_valid_scale_points"],
    n=12,
)
report += "\n\n"

report += "Weakest average log-log curve fits:\n\n"
report += bullet_rows(
    weak_fit_df,
    "channel",
    ["mean_fit_r2", "min_fit_r2", "curve_slope_range", "slope_q0", "mean_valid_scale_points"],
    n=12,
)
report += "\n\n"

high_sep_weak_fit = top_slope_range_df[top_slope_range_df["mean_fit_r2"] < 0.95].copy()

if len(high_sep_weak_fit) > 0:
    report += (
        "Some channels with large q-dependent separation have average fit R² below 0.95. These should be inspected carefully before treating the separation as strong evidence:\n\n"
    )
    report += bullet_rows(
        high_sep_weak_fit,
        "channel",
        ["curve_slope_range", "mean_fit_r2", "min_fit_r2", "slope_qmin", "slope_qmax"],
        n=10,
    )
    report += "\n\n"
else:
    report += (
        "The highest q-dependent separation channels were not flagged below the R²=0.95 reference. This supports the visual interpretation, assuming the individual plots also look smooth.\n\n"
    )

report += (
    "The guiding rule is: curve separation tells us what is interesting; log-log linearity tells us how much confidence to place in that interpretation.\n\n"
)

report += "---\n\n"

report += "## 5. Scale-dependent separation change\n\n"

report += f"- Mean change in low-q versus high-q curve separation from small to large scales: **{fmt(mean_sep_change)}**\n"
report += f"- Median change in low-q versus high-q curve separation from small to large scales: **{fmt(median_sep_change)}**\n\n"

report += f"**Interpretation of median separation change:** {describe_separation_change(median_sep_change)}\n\n"

report += "Largest absolute separation changes:\n\n"
report += bullet_rows(
    top_sep_change_df,
    "channel",
    ["separation_change_from_small_to_large_scale", "first_curve_separation_qmax_qmin_log", "last_curve_separation_qmax_qmin_log", "curve_slope_range", "mean_fit_r2"],
    n=12,
)
report += "\n\n"

report += (
    "Separation change asks whether the distance between low-q and high-q curves grows or shrinks as scale increases. "
    "If separation increases at larger scales, small and large fluctuations become more different at broader time scales. "
    "If separation decreases, the curves become more similar at broader time scales.\n\n"
)

report += "---\n\n"

report += "## 6. Comparison with saved alpha-width values\n\n"

if len(alpha_width_df) > 0:
    report += "Channels with largest saved alpha width:\n\n"
    report += bullet_rows(
        alpha_width_df,
        "channel",
        ["saved_alpha_width", "curve_slope_range", "mean_fit_r2", "slope_q0", "abs_slope_qmax_minus_qmin"],
        n=12,
    )
    report += "\n\n"

    top_alpha = alpha_width_df.iloc[0]

    report += (
        f"The largest saved alpha width is found in **{top_alpha['channel']}** "
        f"(alpha width={fmt(top_alpha['saved_alpha_width'])}, curve slope range={fmt(top_alpha['curve_slope_range'])}, mean R²={fmt(top_alpha['mean_fit_r2'])}).\n\n"
    )

    report += (
        "A strong result is present when saved alpha width and visual q-curve separation agree. "
        "If alpha width is high but q-curve separation is weak or fit quality is low, the channel should be checked carefully.\n\n"
    )

else:
    report += (
        "Saved alpha-width values were not available in the loaded NPZ result objects. The interpretation therefore uses the visual fluctuation-curve summaries computed in this code.\n\n"
    )

report += "---\n\n"

report += "## 7. What makes an all-q fluctuation-curve result stronger\n\n"

report += (
    "A channel-level all-q fluctuation result is stronger when several signs agree:\n\n"
    "- The individual channel plot shows smooth log-log curves.\n"
    "- The curves are approximately straight over the scale range.\n"
    "- The q-dependent slopes differ in an orderly way.\n"
    "- The q curves form a clean fan rather than a jagged bundle.\n"
    "- Low-q and high-q curves separate consistently.\n"
    "- Mean fit R² is high.\n"
    "- The visual pattern agrees with saved alpha width or h(q) summaries.\n"
    "- The contact sheet does not show that the channel is merely an isolated artifact-like outlier.\n\n"
)

report += "A channel-level all-q fluctuation result needs caution when:\n\n"

report += (
    "- Curves are strongly curved on log-log axes.\n"
    "- Curves cross repeatedly.\n"
    "- Only one or two scales dominate the plot.\n"
    "- Low-q or high-q curves are unstable at the endpoints.\n"
    "- Fit R² is low despite large apparent separation.\n"
    "- The channel is visually unlike all neighboring or related channels.\n"
    "- The q-dependent separation is caused by a small number of extreme curves.\n\n"
)

report += "---\n\n"

report += "## 8. Final synthesis\n\n"

if len(top_slope_range_df) > 0:
    row = top_slope_range_df.iloc[0]
    report += (
        f"The strongest q-dependent curve-separation candidate is **{row['channel']}** "
        f"(slope range={fmt(row['curve_slope_range'])}, mean R²={fmt(row['mean_fit_r2'])}). "
    )

if len(top_fit_df) > 0:
    row = top_fit_df.iloc[0]
    report += (
        f"The channel with the strongest average log-log curve linearity is **{row['channel']}** "
        f"(mean R²={fmt(row['mean_fit_r2'])}). "
    )

if len(top_sep_change_df) > 0:
    row = top_sep_change_df.iloc[0]
    report += (
        f"The channel with the largest scale-dependent change in q-curve separation is **{row['channel']}** "
        f"(change={fmt(row['separation_change_from_small_to_large_scale'])}). "
    )

report += "\n\n"

report += (
    "The safest conclusion is not simply that one channel has the most separated curves. "
    "The stronger conclusion is that a channel supports multifractal interpretation when the all-q fluctuation curves are clean, approximately linear, q-dependent in slope, visually coherent, and consistent with the saved MFDFA summary metrics.\n\n"
)

report += "---\n\n"

report += "## 9. One-sentence takeaway\n\n"

report += (
    "The all-q fluctuation plots visually test the foundation of the MFDFA result: reliable multifractal interpretation is strongest when \(F_q(s)\) curves are clean, nearly linear on log-log axes, systematically separated by q, and consistent with the numerical spectrum summaries.\n"
)

with open(report_txt_path, "w") as f:
    f.write(report)

if CAN_DISPLAY_MARKDOWN:
    display(Markdown(report))
else:
    print(report)

print(f"\nSaved interpretation report: {report_txt_path}")
print(f"Done. Individual channel plots are in: {out_dir}")

all_q_fq_interpretation_report = report
all_q_fq_visual_summary_df = summary_df
```
